```python
import pandas as pd
influ = pd.read_csv('/content/drive/MyDrive/instagram_global_influencers_2026.csv')
influ.head()
```

```python
influ.info()
```

## 5 Gráficos con Matplotlib

```python
import matplotlib.pyplot as plt
```

```python
# 1. Histograma: Distribución de Seguidores
plt.figure()
plt.hist(influ['followers'] / 1e6, bins=20, color='skyblue', edgecolor='black')
plt.title('1. Distribución de Seguidores (Matplotlib)')
plt.xlabel('Seguidores (en millones)')
plt.ylabel('Cantidad de Influencers')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# 2. Dispersión: Seguidores vs Likes Promedio
plt.figure()
plt.scatter(influ['followers'] / 1e6, influ['avg_likes'] / 1e6, color='coral', alpha=0.7, edgecolors='black')
plt.title('2. Relación: Seguidores vs Likes Promedio (Matplotlib)')
plt.xlabel('Seguidores (en millones)')
plt.ylabel('Likes Promedio por Post (en millones)')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

```python
# 3. Barras: Top 10 Países
top_countries = influ['country'].value_counts().head(10)
plt.figure()
plt.bar(top_countries.index, top_countries.values, color='lightgreen', edgecolor='black')
plt.title('3. Top 10 Países con más Influencers (Matplotlib)')
plt.xlabel('País')
plt.ylabel('Cantidad de Influencers')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# 4. Boxplot: Distribución de Influence Score
plt.figure()
plt.boxplot(influ['influence_score'], vert=False, patch_artist=True,
            boxprops=dict(facecolor='orchid', color='black'),
            medianprops=dict(color='yellow', linewidth=2))
plt.title('4. Distribución de Puntuación de Influencia (Matplotlib)')
plt.xlabel('Influence Score')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.show()
```

```python
# 5. Dispersión: Posts vs Influence Score
plt.figure()
plt.scatter(influ['posts'], influ['influence_score'], color='gold', alpha=0.7, edgecolors='black')
plt.title('5. Relación: Posts Publicados vs Influence Score (Matplotlib)')
plt.xlabel('Total de Posts')
plt.ylabel('Influence Score')
plt.grid(True, linestyle='--', alpha=0.5)
plt.show()
```

## 5 Gráficos con Seaborn

```python
import seaborn as sns
```

```python
# 1. Histograma: Distribución de Seguidores con KDE
plt.figure(figsize=(10, 6))
sns.histplot(data=influ, x=influ['followers'] / 1e6, bins=20, color='skyblue', kde=True)
plt.title('1. Distribución de Seguidores (Seaborn)')
plt.xlabel('Seguidores (en millones)')
plt.ylabel('Cantidad de Influencers')
plt.show()
```

```python
# 2. Dispersión: Seguidores vs Likes Promedio
plt.figure(figsize=(10, 6))
sns.scatterplot(data=influ, x=influ['followers'] / 1e6, y=influ['avg_likes'] / 1e6, color='coral', alpha=0.8, edgecolor='black')
plt.title('2. Relación: Seguidores vs Likes Promedio (Seaborn)')
plt.xlabel('Seguidores (en millones)')
plt.ylabel('Likes Promedio por Post (en millones)')
plt.show()
```

```python
# 3. Barras: Top 10 Países
top_countries = influ['country'].value_counts().head(10).reset_index()
top_countries.columns = ['country', 'count']

plt.figure(figsize=(10, 6))
sns.barplot(data=top_countries, x='country', y='count', hue='country', palette='viridis', legend=False)
plt.title('3. Top 10 Países con más Influencers (Seaborn)')
plt.xlabel('País')
plt.ylabel('Cantidad de Influencers')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

```python
# 4. Boxplot: Distribución de Influence Score
plt.figure(figsize=(10, 6))
sns.boxplot(data=influ, x='influence_score', color='orchid')
plt.title('4. Distribución de Puntuación de Influencia (Seaborn)')
plt.xlabel('Influence Score')
plt.show()
```

```python
# 5. Dispersión: Posts vs Influence Score con línea de tendencia
plt.figure(figsize=(10, 6))
sns.regplot(data=influ, x='posts', y='influence_score', color='gold', scatter_kws={'edgecolor': 'black', 'alpha': 0.7})
plt.title('5. Relación: Posts Publicados vs Influence Score (Seaborn)')
plt.xlabel('Total de Posts')
plt.ylabel('Influence Score')
plt.show()
```
